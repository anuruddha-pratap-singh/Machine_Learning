```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../Data/tip.csv")
print(df)
```

```python
# df['total_bill'] vs df[['total_bill']]

x = df['total_bill'] # returns a 1D series. use type() to find.
y = df[['total_bill']] # returns a DataFrame.

print(type(x))
print(x.shape)
print(type(y))
print(y.shape)
```

```python
df['tip_percent'] = round(df['tip'] / df['total_bill'] * 100, 2)

df['service'] = 'normal'

df.loc[df['tip_percent'] > 20, 'service'] = 'generous'


sat = df[df['day'] == 'Sat']
print(sat)
```

```python
df.groupby("day").agg(
    Bill_count=("total_bill", 'size'),
    avg_bill=("total_bill", 'mean'),
    avg_tip=('tip', 'mean')
)
```

```python
rows = df.groupby('day')['tip'].idxmax()
print(rows)
```

```python
s = df['day'].value_counts()
print(s)
print(type(s))
```

```python
df = pd.DataFrame(
    {'A': [10, 20, 30], 'B': [40, 50, 60]},
    index=[2, 5, 8]
)
print(df)
print(type(df.loc[5, 'B']))
print(type(df.iloc[1, 0]))
```

```python
df = pd.DataFrame({
    'dept': ['A', 'A', 'B', 'B'],
    'salary': [10, 30, 20, 40]
})

df['avg'] = df.groupby('dept')['salary'].transform('mean')
print(df['avg'].tolist())
```

```python
left = pd.DataFrame({
    'id': [1, 2, 3],
    'name': ['A', 'B', 'C']
})

right = pd.DataFrame({
    'id': [2, 3, 4],
    'marks': [80, 90, 70]
})

result = left.merge(right, on='id', how='inner')
print(result)
```
