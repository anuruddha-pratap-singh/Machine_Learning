```python
A = np.array([[1, 2, 3], [4, 5, 6]])   # (2, 3)
B = np.array([[7, 8, 9]])               # (1, 3)

print(np.vstack((A, B)))  # (3, 3) — rows added
```

```python
C = np.array([[10], [20]])   # (2, 1)
print(np.hstack((A, C)))  # (2, 4) — columns added
```

```python
heights = np.array([170, 155, 180, 165])
weights = np.array([65, 50, 80, 58])

wrong = np.hstack((heights, weights))
#print(wrong)   # (8,)  — just glued end to end!

# Fix: reshape to columns first
X = np.hstack((heights.reshape(-1, 1), weights.reshape(-1, 1)))
print(X)       # (4, 2)  — proper table!
```

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print(np.concatenate((A, B), axis=0))  # same as vstack
print(np.concatenate((A, B), axis=1))  # same as hstack
```

```python
import pandas as pd
restaurant_data = {
    'Table_No': [12, 14, 15, 16 ,17 ,18],
    'Bill_Amount': [1250, 890, 2400,2400,2400,890],
    'Waiter_Name': ['Amit', 'Priya', 'Rahul' ,'anu' , 'pal' ,'pratap']
}
df = pd.DataFrame(restaurant_data)
print(df['Bill_Amount'].value_counts(normalize=True)) 
print()  # count per unique value (sorted desc)
print(df['Bill_Amount'].unique())      
print()   # array of unique values
print(df['Bill_Amount'].nunique())        # number of unique values
```
