## Section 1 — String Validation Methods

**`isalpha()` / `isdigit()` / `isalnum()`**

**Q1.**  A user enters their full name carelessly: `"   rahul  KUMAR sharma  "`
Clean it and print it properly as `Rahul Kumar Sharma`. Also validate that the name contains only alphabets and spaces — print `Invalid name` otherwise.

```python
# Q1
name="   rahul1  KUMAR sharma  "
name=name.strip().title()
lst=name.split()
valid=True

for  i in lst:
    if not i.isalpha():
        valid=False
        break

if valid:
    ans=" ".join(lst)
    print(ans)
else:
    print("Invailid Name")
```

**Q2.**  Marks are given as one messy string: `" 45, 88 ,72,  91 , 60 "`
Convert it into a list of integers, then print the total and average marks.

```python
x=" 45, 88 ,72,  91 , 60 "
lst=x.split(",")
marks=[int(mark.strip()) for mark in lst]
print(sum(marks))
print(sum(marks)/len(marks))
```

## Section 3 — Dictionaries

**Creation + access**
- `d[key]` vs `d.get(key)` — what happens when key is missing?

```python
d={"aman":23,"harsh":27,"Aditi":29,"Mohit":12}
# print(d["aman1"])
print(d.get("aman1","key missing"))
print(d)
```

**`setdefault(key, default)`** — get + insert if missing

```python
print(d.setdefault("aman1",0))

```

```python
print(d)
```

**Views:** `keys()`, `values()`, `items()`

```python
print(d.keys())
print(d.values())
print(d.items())
```

**Update / add:** `update()`, direct assignment

```python
d={"aman":23,"harsh":27,"Aditi":29,"Mohit":12}
d["ankit"]=50
print(d)
d.update({"a":3,"b":6})
print(d)
```

**Looping patterns + membership (`in`)**

```python
d={"aman":23,"harsh":27,"Aditi":29,"Mohit":12}
for k,v in d.items():
    print(k,v)
```
