---
## Step 2 - Separate numeric and categorical columns

```python
nums_col=['cgpa', 'twelfth_percent', 'internships', 'projects', 'backlogs',
       'prep_hours', 'communication_score']

cat_col=['branch','hostel']
```

```python
df.info()
```

---
## Step 3 - Split into X and y

```python
x=df.drop(columns=['student_id','placed','package_lpa'])
y=df['placed']
```

```python
# nums_col=['cgpa', 'twelfth_percent', 'internships', 'projects', 'backlogs',
#        'prep_hours', 'communication_score']

for col in nums_col:
    x[col]=x[col].fillna(x[col].mean())
```

```python
df.isnull().sum()
```

```python
df.isnull().sum()
```

## Step 5 - Encode categorical columns

```python
print(cat_col)
```

```python
df['branch'].value_counts()
```

```python
df['hostel'].value_counts()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder=OneHotEncoder()
```

```python
encoded=encoder.fit_transform(x[cat_col])
print((encoded.shape))

```

```python
print(x.info())
```

```python
print(type(encoded))
```

```python
encoded_df = pd.DataFrame(encoded.toarray(),
                          columns=encoder.get_feature_names_out(cat_col))
encoded_df.head()
```

## Step 6 - Scale numeric columns

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
scaled_col=scaler.fit_transform(x[nums_col])
scaled_col.shape

```

```python
scaled_df = pd.DataFrame(scaled_col, columns=nums_col)
scaled_df.head()
```

```python
scaled_df.describe().round(2)
```

```python
df.describe()
```

---
## Step 7 - Both together with ColumnTransformer

```python
from sklearn.compose import ColumnTransformer
```

```python
preprocessor=ColumnTransformer([
    ('cat',OneHotEncoder(),cat_col),
    ('num',StandardScaler(),nums_col)
]
)
```

```python
final_transform=preprocessor.fit_transform(x)
print(final_transform.shape)
```

```python
preprocessor.get_feature_names_out()
```

```python
final_transform.shape
```

```python
df.head(3)
```

```python
final_df = pd.DataFrame(final_transform,
                        columns=preprocessor.get_feature_names_out())
final_df.head()
```

---
## Step 8 - Train and test split

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    final_transform, y, test_size=0.2, random_state=42)


print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

---
## Step 9 - Train the model

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
model = KNeighborsClassifier(n_neighbors=5)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
y_pred[:10]
```

---
## Step 10 - Score

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test, y_pred)
```
